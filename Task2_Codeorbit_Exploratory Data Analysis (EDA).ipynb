```python
# Create a figure with a defined size
plt.figure(figsize=(10, 5))

# Create a histogram to visualize passenger age distribution
sns.histplot(data=df, x='Age', bins=20, kde=True)

# Add chart title
plt.title('Distribution of Passenger Ages')

# Add X-axis label
plt.xlabel('Age')

# Add Y-axis label
plt.ylabel('Number of Passengers')

# Display the chart
plt.show()
```

```python
# Create a figure
plt.figure(figsize=(10, 5))

# Visualize the distribution of passenger fares
sns.histplot(data=df, x='Fare', bins=30, kde=True)

# Add chart title
plt.title('Distribution of Passenger Fares')

# Label the axes
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')

# Display the chart
plt.show()
```

```python
# Count passengers belonging to each gender
gender_counts = df['Sex'].value_counts()

# Create a bar chart
plt.figure(figsize=(7, 5))
sns.barplot(x=gender_counts.index, y=gender_counts.values)

# Add chart title and axis labels
plt.title('Gender Distribution of Passengers')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

# Display the chart
plt.show()
```

```python
# Count passengers in each passenger class
class_counts = df['Pclass'].value_counts().sort_index()

# Create a bar chart
plt.figure(figsize=(7, 5))
sns.barplot(x=class_counts.index, y=class_counts.values)

# Add chart title and axis labels
plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

# Display the chart
plt.show()
```

```python
# Count passengers based on their survival status
survival_counts = df['Survived'].value_counts().sort_index()

# Create a bar chart
plt.figure(figsize=(7, 5))
sns.barplot(x=survival_counts.index, y=survival_counts.values)

# Add chart title and axis labels
plt.title('Survival Distribution of Passengers')
plt.xlabel('Survival Status (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

# Display the chart
plt.show()
```

```python
# Compare passenger survival across genders
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Sex', hue='Survived')

# Add chart title and labels
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

# Display legend and chart
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
# Compare survival counts across passenger classes
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Pclass', hue='Survived')

# Add chart title and labels
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

# Display chart
plt.show()
```

```python
# Create a scatter plot to examine the relationship between age and fare
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')

# Add chart title and labels
plt.title('Relationship Between Passenger Age and Fare')
plt.xlabel('Age')
plt.ylabel('Fare')

# Display chart
plt.show()
```

```python
# Compare age distributions between survivors and non-survivors
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Survived', y='Age')

# Add chart title and labels
plt.title('Age Distribution by Survival Status')
plt.xlabel('Survival Status (0 = No, 1 = Yes)')
plt.ylabel('Age')

# Display chart
plt.show()
```

```python
# Select numerical columns for correlation analysis
numerical_df = df.select_dtypes(include=np.number)

# Calculate correlation between numerical variables
correlation_matrix = numerical_df.corr()

# Create a heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

# Add chart title
plt.title('Correlation Heatmap of Titanic Dataset')

# Display chart
plt.show()
```

```python
# Create a box plot to identify unusual fare values
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y='Fare')

# Add chart title
plt.title('Outlier Detection in Passenger Fare')

# Display chart
plt.show()
```

```python
# Create a box plot to identify unusual age values
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y='Age')

# Add chart title
plt.title('Outlier Detection in Passenger Age')

# Display chart
plt.show()
```

```python
# Calculate survival percentage for each gender
gender_survival = pd.crosstab(
    df['Sex'],
    df['Survived'],
    normalize='index'
) * 100

# Display the results
print("Survival Percentage by Gender:")
display(gender_survival.round(2))
```

```python
# Calculate the percentage of passengers who survived and did not survive
survival_percentage = df['Survived'].value_counts(normalize=True) * 100

# Display survival percentages
print("Survival Percentage:")
print(survival_percentage.round(2))
```

```python
# Calculate survival percentage for each passenger class
class_survival = pd.crosstab(
    df['Pclass'],
    df['Survived'],
    normalize='index'
) * 100

# Display the results
print("Survival Percentage by Passenger Class:")
display(class_survival.round(2))
```
